```python
import numpy as np 
import pandas as pd
```

```python
#load database
df = pd.read_csv("amazon.csv")

print("Amazon Product Data Analyzer")
print("-" * 40)
print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
df.select_dtypes(include=["object", "str"]).columns
#clan cloum name
df.columns = df.columns.str.strip().str.lower()
# Remove duplicate records
df = df.drop_duplicates().copy()

# Convert relevant columns to numeric values
numeric_columns = [
    "actual_price",
    "discounted_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = (
            df[column]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("₹", "", regex=False)
            .str.replace("$", "", regex=False)
            .str.replace("%", "", regex=False)
            .str.strip()
        )

        df[column] = pd.to_numeric(
            df[column], errors="coerce"
        )

# Clean text columns
for column in df.select_dtypes(include=["object", "str"]).columns:
    df[column] = df[column].str.strip()
    df[column] = df[column].str.strip()

print("Data cleaning completed!")
print("Dataset shape:", df.shape)
```

```python
df['Savings'] = df['actual_price'] - df['discounted_price']
print("PRICE ANALYSIS")
print("----------------------------------------")
print("Average actual price:")
print(df["actual_price"].mean())
print("Average discounted price:")
print(df["actual_price"].mean()-df["discount_percentage"].mean())
print("Average savings:")
print(df["discount_percentage"].mean())

top_10=df.nlargest(10, 'Savings')
print(top_10[
    ['product_name', 'actual_price',
     'discounted_price', 'Savings']
].to_string(index=False))
```

```python
print("CATEGORY ANALYSIS")
print("-"*10)
print("Number of products by category:")
category_counts = df.groupby('category')['product_id'].nunique().reset_index(name='product_count')
print(category_counts)
avg_rating = df.groupby('category')['rating'].mean().reset_index(name='avg_rating')
print("Average rating by category:")
print(df.groupby('category')['rating'].mean())
```

```python
print("CORRELATION ANALYSIS")
print("-"*100)
numeric_data=df.select_dtypes(include="number")
if numeric_data.shape[1] >= 2:
    display(numeric_data.corr().round(2))
else:
    print("Not enough numeric columns for correlation.")
```

```python
print("OUTLIER DETECTION")
print("-" * 40)

columns = [
    "actual_price",
    "discounted_price",
    "rating"
]

for column in columns:

    if column not in df.columns:
        continue

    values = df[column].dropna()

    if len(values) < 4:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = df[
        (df[column] < lower_limit)
        | (df[column] > upper_limit)
    ]

    print(f"\nColumn: {column}")
    print("Lower limit:", round(lower_limit, 2))
    print("Upper limit:", round(upper_limit, 2))
    print("Number of outliers:", len(outliers))
```
